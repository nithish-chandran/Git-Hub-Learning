```python
# addition of 2 numbers
a=int(input("Enter first number: "))
b=int(input("Enter second number: "))

add=lambda x,y:x+y
add (a,b)
```

```python
# create a function for finding the area of the rectangular plot

length=int(input("Enter the Length: "))
breadth=int(input("Enter the breadth: "))

area=lambda length, breadth: length*breadth
area (length,breadth)
```

```python
# create a function to check if a number is even or not

num=int(input("Enter the number: "))

Even_number=lambda num: "The number is even" if num%2==0 else "The number is odd"

Even_number(num)
```

```python
# map()  filter()   reduce()

#syntax
map(function,iterable)
```

```python


# to find the square of each element
list_numbers=[10,23,45,84,100]
square=list(map(lambda x:x**2, list_numbers))
print(square)
```

```python
# you have a salaries of employees, update the salaries by 10%

Salaries=[25000,40000,10000,30000,55000]
Increment=list(map(lambda a:int(a*0.1+a), Salaries))
print(Increment)
```

```python
# you have names of customers in lowercase letter, capitalize each customer's name
# and print the updated customer names

Customer_Name=["Nithish", "Veeresh", "Naveen", "Sohan"]
Updated_Name=list(map(lambda x:x.upper(),Customer_Name))
print(Updated_Name)
```

```python
# filter()
#syntax:

# filter(function,iterable)
# used to filter out elements from the iterable
```

```python
# filter out marks of students who have passed the exam

Marks=[55,78,21,90,45,33]
Students_Passed=list(filter(lambda x:x>35, Marks))

print(Students_Passed)
                         
```

```python
# filter out products that are available in stock

Products={"Laptop": 5,"Bag": 0,"Mouse": 20,"Keyboard": 0,"Charger": 12}
Available_Products=list(filter(lambda x:x[1]>0,Products.items()))
print(Available_Products)
```

```python
Products={"Laptop": 5,"Bag": 0,"Mouse": 20,"Keyboard": 0,"Charger": 12}
Available_Products=list(map(lambda x:x[0],(filter(lambda x:x[1]>0,Products.items()))))
print(Available_Products)
```

## Reduce

```python
#syntax:
#reduce(function,iterable)

#functools

from functools import reduce 
marks=[55,78,21,90,45,33]
total_marks=reduce(lambda x,y:x+y,marks)
print(total_marks)
```

```python
marks=[55,78,21,90,45,33]
total_marks=reduce(lambda x,y:x+y,marks)
average=total_marks/len(marks)                  #we can also use x,y:x+y/2
print(average)
```

```python
marks=[55,78,21,90,45,33]
total_marks=reduce(lambda x,y:(x+y)/2,marks)

print(total_marks)
```
