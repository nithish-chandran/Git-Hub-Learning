```python
# A structured way to store, organize and manage the dat
# Types: List, Tuple, Set, Dictionary

        # Collection of items. denoted using []. 
        # Ordered - every item as a particular index
        # ex: list1 = [1,2,3,4,5]

# hold multiple/different data types
# eg: items=[70, 86.5,"fruits", "stationaries", True]
# it allows duplicate values

# list2=["a", "a", "b", "c", "c", "d"]

# list:
age=[21,23,34,67]
weight=[45.7,65.2,98.3]
items=['Fruits', 'Vegetables', 'Stationaries']
my_list=[12,34,67.5,True,'Total']

# list () : built-in funtion to create a list and converts other data structures
num=list([2,3,4,5])

print(age)
print(items)
```

```python
# can access elements using index
items=['Fruits', 'Vegetables', 'Stationaries']
print(items[0])
print(items [-1])
```

```python
# slicing
my_list=[12,34,67.5,True,'Total']
print(my_list[1:4:2])
```

```python
# mutable - we can change the value in the list. add or delete items in the list
```

```python
# append(): will add a new item at the end of your list
my_list.append(1000)
print(my_list) 
items.append("Clothes")
print(items)
```

```python
# insert(): adds a new itens in a specific position
# syntax: insert(index, item)
items.insert(2,'Grains')
print(items)
my_list.insert(4,4000)
print(my_list)
```

```python
# extend(): used to add multiple items to your list. Will take the iterable. It will be added to the end
# syntax: list_name.extend(iterable)
items.extend([100,200,300])
print(items)
my_list.extend('abcd')
print(my_list)  
my_list.extend(['arnab', 'bc', 'cd'])
print(my_list)
```

```python
# replacing items in a list

print(items)
items[1]='Electricity'
items[5]='Water'
print(items)
```

```python
# removing items from the list

# 1. pop(): remove an item from the list using index values

print(my_list)
my_list.pop(5)
print(my_list)
```

```python
# in default pop() deletes the last item from the list

my_list.pop()
print(my_list)
```

```python
# remove(): deletes the item by its value

my_list.remove(True)
print(my_list)
```

```python
# remove() deletes the first occured item from the list
my_list.remove(1000)
print(my_list)
```

```python
# list concatenation

l1=[1,2,3,4]
l2=[3,4,5,6,7]
l3=["apple","orange","mango"]
join_list=l1+l2+l3
print(join_list)
```

```python
fruits=['mango', 'orange', 'grapes', 'apple']
fruits.sort()
print(fruits)
```

```python
# this is used to sort in descending order

fruits.sort(reverse=True)
print(fruits)
```

```python
# false values:

# 0, false, none, empty values (" ")

# bool (0.0) = false, bool () = false, bool (none) = false

# learn about all() & any()
#all()
#Returns True if all elements in the iterable are True.
#Returns False if any element is False.
```

```python
l2=[3,4,5,6,7]
print(l2)
```

```python
print(all(l2))
```

```python
l3=[1,2,3,4,0]
print(all(l3))
```

```python
# any()

# Returns True if at least one element in the iterable is True.
# Returns False if all elements are False.

print(any(l3))
```

```python
l4=[0,False,None]
print(any(l4))
```

```python
l5=[1,0,False,' ']
print(any(l5))
```
