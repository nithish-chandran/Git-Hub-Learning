## Set

```python
# Sets are represented using {}
# They are unordered (means we cannot access the item using index)
# Sets cannot have duplicates. only Unique elements are allowed
# Sets are mutable (we can change the items)
# Holds different data types

eg: Set1={10,12,14,16}
    Set2={'a','b','c','d'}
    Set3={'sports', 'music','film'}

# Set () We can create a set and convert other data structures to set
```

```python
set_1={100,200,300,400}
print(set_1)
```

```python
set_2=set([12,13,14])  #converting a list into a set using set()
print(set_2)
```

```python
set_2[0]
```

```python
# add(): used to add a single item to your set

set_1.add(500)
print(set_1)
```

```python
set_1.add(500)                   #in set when adding this 500 can be added anywhere
print(set_1)
```

```python
# update(iterable) : to add multiple items to the set

set_1.update([600,700,800,900])
print(set_1)
```

```python
set_2.update([25,26,27])
print(set_2)
```

```python
set_1.remove(700)
print(set_1)
```

```python
# discard(item) : another method for removing

set_1.discard(600)
print(set_1)
```

```python
set_1.discard(600)         # means even if the item is not availabele.. it will not through an error. We can continue the code
print(set_1)
```

```python
# remove (item) : remove item from the set

set_1.remove(700)
print(set_1)
```

```python
# remove() : using remove method if a particular item is not present in the set, it will throw an error and disrupt the flow of your program

# discard() : if item does not exists, it will not throw an error
```

```python
# pop() : delete random items

set_1.pop()
print(set_1)
```

```python
set_1.clear()
print(set_1)
```

```python
del set_1
```

```python
# union()                   - is to combine all elements of the sets
# intersection()            - Will give the common elements between the set
# difference()              - Will give the different elements between the sets
```

```python
a={1,2,3,4}
b={3,4,5,6}
c={7,8}
```

```python
a.union(b)
```

```python
a.union(b,c)
```

```python
b.union(a)
```

```python
a.intersection(b)
```

```python
a.intersection(b,c)   # if no common element, it will return the empty set
```

```python
a.difference(b)         # a-b
```

```python
b.difference(a)         # b-a
```

## Loop

```python
# used to repeat a block of code multiple times
# for loop            - used when we know number of times the code should be repeated
# while loop 

# for loop:
# syntax
for variable in sequence: 
    # block of codes
# sequence - collection of items
# variable - represents each item inside the sequence

# eg: print numbers from 1 to 10

# range() - to generate a sequence of numbers. By default, the range starts from 0
# range (10) 0.1.2.3.....,9
# range (1,11) 1,2,3,...10



         - 

```

```python
# print numbers from 1 - 20
for i in range(1,21):
    print(i)
```

```python
for i in range(1,21):
    print(i,end=" ")         # to print in a vertical order
```

```python
# print every item in the list
items=['fruits','vegetables','stationaries','electricity','water']

for i in items:
    print(i)
    
```

```python
for i in items:
    print(i,end=" ")
```

```python
## loop control statements
# break                    - it will exit the loop immediately 
# continue

# eg for break

for i in items:
    if i=='electricity':
        break
    print(i)
```

```python
# continue - it will skip the current iteration & continue with the rest

for i in items:
    if i=='stationaries':
        continue
    print(i)
```

```python
for i in items:
    if i=='stationaries' or i=='electricity':
        continue
    print(i)
```

```python
# print the square of each number from a list

sqr=[1,2,3,4,5]
for q in sqr:
    print(q**2)
```

```python
# nested for loop
# having a for loop inside another for loop
# syntax:
for variable1 in sequence1:
    for variable2 in sequence1:
        # block of code
```

```python
items=['apple', 'banana', 'mango', 'orange']
price=[60,30,75,50]

for i in items:
    for p in price:
        print(i,p)
```

```python
for i in items:
    for p in price:
        if items.index(i)==price.index(p):
            print(i,p)
```
