```python
# datetime module
# to create dates using date(yyyy,mm,dd)
from datetime import date

date_1=date(2002,1,18)
print(date_1)
```

```python
# to get today's date
# date.today()
today_1=date.today()
print(today_1)
```

```python
# isoweekday() : to get the weeday in numbers starting from 1 to 7

today_1.isoweekday()
```

```python
# to create time using time(hh,mm,ss)
from datetime import time
time_1=time(12,30,45)
print (time_1)
```

```python
# to deal with date as well as time

from datetime import datetime
current=datetime.now()               # to get today's date and current time
print(current)
time_1.hour
time_1.minute
```

```python
from datetime import timedelta as td
print(today_1)
future_date = today_1 + td(days=730)
print(future_date)
```

```python
from datetime import timedelta as td
print(today_1)
future_date = today_1 + td(weeks=7)          # for weeks
print(future_date)
```

```python
future_date.month
```
